## Q7. Sum of weights

```python
asia = df[df["origin"] == "Asia"]
```

```python
selected_columns = asia[["vehicle_weight","model_year"]]
```

```python
limited = selected_columns.head(7)
```

```python
X = np.array(limited)
```

```python
XTX = X.T.dot(X)
```

```python
inverse = np.linalg.inv(XTX)
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
# Multiply the inverse of XTX with the transpose of X, and then multiply the result by y
w = inverse.dot(X.T).dot(y)
```

```python
w.sum()
```
